```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# 设置全局字体为 Arial Unicode MS（macOS 系统自带）
plt.rcParams['font.sans-serif'] = ['Arial Unicode MS'] 
# 解决坐标轴负号显示为方块的问题
plt.rcParams['axes.unicode_minus'] = False 
```

```python
x = np.linspace(0, 2 * np.pi, 200)
y = np.sin(x)

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(x, y)
ax.set_xlabel("x")
ax.set_ylabel("sin(x)")
ax.set_title("正弦曲线")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 2 * np.pi, 200)
y1, y2 = np.sin(x), np.cos(x)

fig, ax = plt.subplots(figsize=(7, 4), constrained_layout=True)

ax.plot(x, y1, label="sin(x)", color="tab:blue", linewidth=2)
ax.plot(x, y2, label="cos(x)", color="tab:orange", linestyle="--")

ax.set_xlabel("x / rad")
ax.set_ylabel("y")
ax.set_title("正弦与余弦")
ax.legend(loc="upper right")
ax.grid(True, alpha=0.3)

fig.savefig("sin_cos.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.animation import FuncAnimation

fig, ax = plt.subplots()
x = np.linspace(0, 2 * np.pi, 200)
line, = ax.plot(x, np.sin(x))
ax.set_ylim(-1.2, 1.2)

def update(frame):
    line.set_ydata(np.sin(x + frame / 10))
    return line,

ani = FuncAnimation(fig, update, frames=100, interval=50, blit=True)
ani.save("wave.gif", writer="pillow", fps=20)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

fig = plt.figure(figsize=(7, 5))
ax = fig.add_subplot(projection="3d")

x = np.linspace(-5, 5, 50)
y = np.linspace(-5, 5, 50)
X, Y = np.meshgrid(x, y)
Z = np.sin(np.sqrt(X**2 + Y**2))

ax.plot_surface(X, Y, Z, cmap="viridis")
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")
plt.show()
```
