### In Brief: Tensor Broadcasting

```python
rand = torch.rand(2, 4)
doubled = rand * (torch.ones(1, 4) * 2)
# (2, 4)
# (1, 4)

print(rand)
print(doubled)
```

The rules for broadcasting are:

- Each tensor must have at least one dimension - no empty tensors.

- Comparing the dimension sizes of the two tensors, going from last to first:

- Each dimension must be equal, or

- One of the dimensions must be of size 1, or

- The dimension does not exist in one of the tensors

```python
a = torch.ones(4, 3, 2)

b = a * torch.rand(3, 2)  # (,3,2)
print(b)

c = a * torch.rand(3, 1)  # (,3,1)
print(c)

d = a * torch.rand(1, 2)  # (,1,2)
print(d)
```

```python
a =     torch.ones(4, 3, 2)

b = a * torch.rand(4, 3)    # dimensions must match last-to-first

c = a * torch.rand(   2, 3) # both 3rd & 2nd dims different

d = a * torch.rand((0, ))   # can't broadcast with an empty tensor
```

### More Math with Tensors

```python
# common functions
a = torch.rand(2, 4) * 2 - 1
print("Common functions:")
print(torch.abs(a))
print(torch.ceil(a))
print(torch.floor(a))
print(torch.clamp(a, -0.5, 0.5))

# trigonometric functions and their inverses
angles = torch.tensor([0, math.pi / 4, math.pi / 2, 3 * math.pi / 4])
sines = torch.sin(angles)
inverses = torch.asin(sines)
print("\nSine and arcsine:")
print(angles)
print(sines)
print(inverses)

# bitwise operations
print("\nBitwise XOR:")
b = torch.tensor([1, 5, 11])
c = torch.tensor([2, 7, 10])
print(torch.bitwise_xor(b, c))

# comparisons:
print("\nBroadcasted, element-wise equality comparion:")
d = torch.tensor([[1.0, 2.0], [3.0, 4.0]])
e = torch.ones(1, 2)
print(torch.eq(d, e))

# reductions:
print("\nReduction ops:")
print(torch.max(d))
print(torch.max(d).item())
print(torch.mean(d))
print(torch.std(d))
print(torch.prod(d))
print(torch.unique(torch.tensor([1, 2, 1, 2, 1, 2])))

# vector and linear algebra operations
v1 = torch.tensor([1.0, 0.0, 0.0])
v2 = torch.tensor([0.0, 1.0, 0.0])
m1 = torch.rand(2, 2)
m2 = torch.tensor([[3.0, 0.0], [0.0, 3.0]])
print("\nVectors & Matrices:")
print(torch.linalg.cross(v2, v1))
print(m1)
m3 = torch.matmul(m1, m2)
print(m3)
print(torch.linalg.svd(m3))
```

### Altering Tensors in Place

```python
a = torch.tensor([0, math.pi / 4, math.pi / 2, 3 * math.pi / 4])
print(f"{a=}")
print(torch.sin(a))
print(f"{a=}")

print("")

b = torch.tensor([0, math.pi / 4, math.pi / 2, 3 * math.pi / 4])
print(f"{b=}")
print(b.sin_())
print(f"{b=}")
```

```python
a = torch.ones(2, 2)
b = torch.rand(2, 2)

print("Before:")
print(f"{a=}")
print(f"{b=}")

print("\nAfter add_")
print(f"{a.add_(b)=}")
print(f"{a=}")
print(f"{b=}")

print("\nAfter multiplying")
print(f"{b.mul_(b)=}")
print(f"{b=}")
```

```python
a = torch.rand(2, 2)
b = torch.rand(2, 2)
c = torch.zeros(2, 2)
old_id = id(c)

print(f"{c=}")
d = torch.matmul(a, b, out=c)
print(f"{c=}")

assert c is d
assert id(c) == old_id

torch.rand(2, 2, out=c)
print(f"{c=}")
assert id(c) == old_id
```

## Copying Tensors

```python
a = torch.ones(2, 2)
b = a

a[0][1] = 561
print(f"{b=}")
```

```python
a = torch.ones(2, 2)
b = a.clone()

assert b is not a
print(torch.eq(a, b))

a[0][1] = 561
print(f"{b=}")
```

```python
a = torch.rand(2, 2, requires_grad=True)  # turn on autograd
print(a)

b = a.clone()
print(b)

c = a.detach().clone()
print(c)

print(a)
```

```python
a = torch.tensor([1.0, 2.0, 3.0], requires_grad=True)
b = a.clone()  # requires_grad 继承了 True

# 用 b 参与计算
loss = (b * 2).sum()
loss.backward()

print(f"a.grad: {a.grad}")  # 输出: tensor([2., 2., 2.]) ✅ 有值，且正确
print(f"b.grad: {b.grad}")  # 输出: None ❌ 没有值（非叶子节点被丢弃）
```

```python
import torch

a = torch.rand(2, 2, requires_grad=True)
# 假设 a 已经有了一些梯度（比如反向传播后）
a.sum().backward()  # 让 a.grad 有值（全1矩阵）

# 完全隔离复制
b = a.detach().clone().requires_grad_(True)

# 1. 数值隔离（修改 b 不影响 a）
with torch.no_grad():
    b[0, 0] = 999.0  # ✅ 完全不影响 a，且不报错

print(f"{a=}")  # 输出还是原来的随机数，不受影响 ✅
print(f"{b=}")

# 2. 梯度内存隔离（反向传播各走各的）
loss_b = b.sum()
loss_b.backward()  # 梯度流向 b
print(f"{a.grad=}")      # 依然是之前剩下的全1矩阵，没有叠加新梯度 ✅ 完全隔离
print(f"{b.grad=}")      # 输出: tensor([[1., 1.], [1., 1.]]) ✅ 有值
print(f"{b.grad is a.grad=}")  # False ✅
```
